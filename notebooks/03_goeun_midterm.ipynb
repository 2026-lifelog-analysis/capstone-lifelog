## 1. 개인 vs 집단 기준선 (라벨 없음)
- **같은 날 비율**: 개인 기준선이 울린 날 중 집단 기준선도 울린 비율. 낮을수록 두 방법이 '다른 날'을 잡음
- **0번인 사람**: 경보가 한 번도 안 울린 사람 수
- **상위 10% 몫**: 경보를 가장 많이 받은 10% 사람이 전체 경보에서 차지하는 비율 (높을수록 몇 사람에게 몰림)
- **사람별 오경보율**: 라벨이 없으니 모든 경보를 오경보로 볼 때 사람별 경보율 (중앙값 [1사분위, 3사분위])

```python
def compare_table(dow_adjust=False):
    rows = []
    for c in ['steps', 'sleep_min', 'bed_min']:
        d = nh.assign(p=bl.bad_score(bl.personal_z(nh, c, dow_adjust=dow_adjust), c),
                      g=bl.bad_score(bl.group_z(nh, c, min_n=30), c))
        r = ev.compare_alarms(d, 'p', 'g')
        fa = [f'{m:.3f} [{q1:.3f}, {q3:.3f}]' for m, q1, q3 in r['사람별오경보율']]
        rows.append({'지표': c, '같은 날 비율': round(r['같은날비율'], 3),
                     '0번인 사람 (개인/집단)': r['0번인사람'],
                     '상위 10% 몫 (개인/집단)': tuple(round(v, 3) for v in r['상위10%몫']),
                     '사람별 오경보율 개인': fa[0], '사람별 오경보율 집단': fa[1],
                     '사람 수': r['사람수'], '_raw': r})
    return pd.DataFrame(rows)

cmp = compare_table(dow_adjust=False)
cmp.drop(columns='_raw')
```

요일 보정을 켜면 (요일 평균은 기준선 창 안의 과거 값으로만 계산):

```python
cmp_dow = compare_table(dow_adjust=True)
cmp_dow.drop(columns='_raw')
```

## 2. 가짜 이탈 주입 평가

사람마다 '정상 구간'(직전 28일에 기록 14일 이상, 구간 안에 기록 7일 이상) 하나를 무작위로 골라 **14일** 동안 지표를 바꿈.
- **탐지**: 주입 구간 14일 안에 하루라도 경보 → 탐지율 (주 지표)
- **대조군**: 같은 구간을 주입 없이 돌렸을 때 울리면 오경보 → 오경보율
- **차이** = 탐지율 − 오경보율 (Youden's J). 0이면 우연 수준
- **탐지 지연**: 주입 시작 후 첫 경보까지 며칠 (0 = 첫날)
- 기준값: 대조군 전체에서 하루 오경보율 3.3%가 되도록 방법마다 하나. CUSUM은 경보선 h를 그렇게 맞춤
- 반복 10번(시드 0~9)으로 사람마다 비율을 낸 뒤, 표의 숫자는 **사람별 값의 중앙값**
- PR-AUC는 보조 (주입 구간 날 vs 같은 구간 대조군 날)

비교 방법: 무작위 / 단일 z(|z|) / 개인 기준선 / 개인 기준선 + CUSUM / 집단 기준선

```python
SCENARIOS = [
    # (지표, 크기, 'relative'=비율 | 'absolute'=더하기, 모양)
    ('steps',   -0.3, 'relative', 'sudden'),    # 걸음 30% 감소, 갑자기
    ('steps',   -0.3, 'relative', 'gradual'),   # 걸음 30% 감소, 서서히
    ('bed_min', 120,  'absolute', 'sudden'),    # 잠든 시각 2시간 늦어짐, 갑자기
    ('bed_min', 120,  'absolute', 'gradual'),   # 잠든 시각 2시간 늦어짐, 서서히
]
N_REPS, SEED = 10, 0

results = {}
for col, eff, mode, shape in SCENARIOS:
    key = f'{col} {eff:+g} {shape}'
    results[key] = ev.evaluate_injection(nh, col, eff, mode=mode, shape=shape, n_reps=N_REPS, seed=SEED)
    print('완료:', key, '| CUSUM h =', round(results[key]['setting']['cusum_h'], 2))
```

```python
for key, res in results.items():
    print(f'■ {key}')
    display(ev.results_table(res))
```

## 3. 시험 기간 비교 (개인 vs 집단 기준선)

가설: 시험 주에는 모든 학생의 생활이 같이 흔들림. 이때
- **개인 기준선**은 "이 학생이 평소보다 달라짐"을 잡아 경보가 늘어나고
- **집단 기준선**은 "다들 그러니 정상"으로 보고 조용할 것

설정
- 지표: steps, sleep_min, bed_min. 개인 = `bad_score(personal_z)`, 집단 = `bad_score(group_z, min_n=30)`
- 경보 기준값: 1장과 같이 **두 점수가 모두 있는 날 전체**에서 상위 1/30 (`ev.ALARM_RATE`), 방법마다 하나
- 달력 범위 밖이라 event_tag/term_tag가 빈 날은 기준값 계산에는 들어가지만 표에서는 빠짐
- 결과는 기록하지 않음 (`log_result` 안 씀)

```python
EX_COLS = ['steps', 'sleep_min', 'bed_min']
EVENTS = ['시험', '휴일', '개강주', '없음']
TERMS = ['학기', '방학']

ex = {}
for c in EX_COLS:
    d = nh[['pid', 'date', 'term_tag', 'event_tag', c]].assign(
        p=bl.bad_score(bl.personal_z(nh, c), c),
        g=bl.bad_score(bl.group_z(nh, c, min_n=30), c)).dropna(subset=['p', 'g'])
    d['p_alarm'] = d['p'] > d['p'].quantile(1 - ev.ALARM_RATE)   # compare_alarms와 같은 기준
    d['g_alarm'] = d['g'] > d['g'].quantile(1 - ev.ALARM_RATE)
    ex[c] = d
    print(f"{c}: 날 수 {len(d)}, 전체 경보율 개인 {d['p_alarm'].mean():.3%} / 집단 {d['g_alarm'].mean():.3%}")
```

```python
def tag_rate_table(tag_col, order):
    rows = []
    for c, d in ex.items():
        for t in order:
            x = d[d[tag_col] == t]
            rows.append({'지표': c, tag_col: t,
                         '개인 경보율': round(x['p_alarm'].mean(), 4),
                         '집단 경보율': round(x['g_alarm'].mean(), 4),
                         '날 수': len(x), '사람 수': x['pid'].nunique(),
                         '원래 값 중앙값': x[c].median()})
    return pd.DataFrame(rows)

ev_tab = tag_rate_table('event_tag', EVENTS)
term_tab = tag_rate_table('term_tag', TERMS)
print('■ event_tag별 (원래 값 중앙값: steps=걸음, sleep_min=분, bed_min=정오부터 분)')
display(ev_tab)
print('■ term_tag별')
display(term_tab)
```

```python
C_P, C_G = '#2a78d6', '#eb6834'   # 개인 = 파랑, 집단 = 주황
fig, axes = plt.subplots(1, 3, figsize=(15, 4), sharey=True)
xs, w = np.arange(len(EVENTS)), 0.38
for ax, c in zip(axes, EX_COLS):
    t = ev_tab[ev_tab['지표'] == c].set_index('event_tag').loc[EVENTS]
    ax.bar(xs - w / 2 - 0.01, t['개인 경보율'] * 100, w, color=C_P, label='개인 기준선')
    ax.bar(xs + w / 2 + 0.01, t['집단 경보율'] * 100, w, color=C_G, label='집단 기준선')
    ax.axhline(ev.ALARM_RATE * 100, color='gray', ls='--', lw=1, label='기준 3.3%')
    ax.set_xticks(xs, EVENTS); ax.set_title(c)
    ax.grid(axis='y', alpha=0.3); ax.set_axisbelow(True)
    ax.spines[['top', 'right']].set_visible(False)
axes[0].set_ylabel('경보율 (%)')
axes[0].legend(frameon=False)
plt.tight_layout(); plt.show()
```

### (참고용) 사람 단위: 시험 주 경보율 − 평소(없음) 경보율
- 사람마다 (시험 날 경보율) − (없음 날 경보율)을 개인/집단 각각 구함
- 시험 날과 없음 날이 각각 `MIN_TAG_DAYS`(5)일 이상인 사람만 (하루 이틀로 0%/100%가 나오는 것 방지)
- Wilcoxon: 같은 사람끼리 짝지어 개인 차이 vs 집단 차이 비교 (차이가 0인 사람은 검정에서 빠짐)
- ⚠️ **주의**: 대부분 학생은 시험 주 경보가 0번임 (시험 날은 사람당 중앙값 17일, 지표·방법별로 53~72%가 0번). 그래서 차이 중앙값이 음수로 쏠리므로 이 표는 참고용으로만 보고, 결론은 위 event_tag별 경보율 표 기준으로 씀

```python
from scipy.stats import wilcoxon

MIN_TAG_DAYS = 5
rows = []
for c, d in ex.items():
    x = d[d['event_tag'].isin(['시험', '없음'])]
    g = x.groupby(['pid', 'event_tag'])
    r = g[['p_alarm', 'g_alarm']].mean().unstack('event_tag')
    n = g.size().unstack('event_tag')
    ok = (n.get('시험', 0) >= MIN_TAG_DAYS) & (n.get('없음', 0) >= MIN_TAG_DAYS)
    dp = (r[('p_alarm', '시험')] - r[('p_alarm', '없음')])[ok]
    dg = (r[('g_alarm', '시험')] - r[('g_alarm', '없음')])[ok]
    rows.append({'지표': c, '사람 수': int(ok.sum()),
                 '개인 차이 중앙값': round(dp.median(), 4), '집단 차이 중앙값': round(dg.median(), 4),
                 '개인 차이 > 0 비율': round((dp > 0).mean(), 3), '집단 차이 > 0 비율': round((dg > 0).mean(), 3),
                 'Wilcoxon p (개인 vs 집단)': float(wilcoxon(dp, dg).pvalue) if (dp != dg).any() else np.nan})
exam_person = pd.DataFrame(rows)
exam_person
```

**결과 해석** (event_tag별 경보율 표 기준)
- 개인 기준선은 울리고 집단 기준선은 조용한, 가설과 같은 모양이 세 곳에서 보임
  - **휴일 걸음**: 개인 14.5% vs 집단 1.6%. 다들 덜 걷는 날이라 집단은 거의 안 울리고, 개인 기준선의 걸음 경보는 휴일에 몰림
  - **개강주 잠든 시각**: 개인 5.3% vs 집단 2.5%
  - **시험주 수면**: 개인 4.3% vs 집단 3.0%. 수면 중앙값도 416분으로 평소(426분)보다 짧음
- 시험주 자체로는 수면만 가설에 맞음. 걸음은 개인 2.2%(평소 2.3%)인데 집단이 4.9%로 더 울리고, 잠든 시각도 개인 3.5% < 집단 4.0%
- 사람 단위 시험 주 − 평소 차이(위 표)는 **참고용**: 대부분 학생의 시험 주 경보가 0번이라 중앙값이 음수로 쏠림
